```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from textblob import TextBlob


# Load the uploaded file (no change needed if it's named twitter_training.csv)
df = pd.read_csv('twitter_training.csv', header=None)
df.columns = ['User', 'Sentiment', 'Tweet_ID', 'Tweet']
df = df[['Sentiment', 'Tweet']]

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    # Convert the input to string to handle potential non-string entries
    text = str(text)
    text = text.lower()
    text = re.sub(r'http\\S+|www\\S+|https\\S+', '', text)
    text = re.sub(r'[^a-zA-Z\\s]', '', text)
    # Use TextBlob for tokenization instead of nltk.word_tokenize
    tokens = TextBlob(text).words
    tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]
    return ' '.join(tokens)

df['Cleaned_Tweet'] = df['Tweet'].apply(preprocess_text)

df.head()
```

```python
# Function to get VADER sentiment scores
def get_vader_sentiment_scores(text):
    if isinstance(text, str):
        return analyzer.polarity_scores(text)
    else:
        return {'neg': 0.0, 'neu': 0.0, 'pos': 0.0, 'compound': 0.0}

# Apply VADER sentiment analysis to the cleaned tweets
df['Vader_Sentiment'] = df['Cleaned_Tweet'].apply(get_vader_sentiment_scores)

# Extract compound score into a separate column
df['Vader_Compound_Score'] = df['Vader_Sentiment'].apply(lambda score_dict: score_dict['compound'])

display(df.head())
```

## Compare textblob and vader

### Subtask:
Calculate sentiment scores using TextBlob. Compare TextBlob scores with VADER scores and the original sentiment labels. Assess the agreement or differences between the two sentiment analysis tools.


**Reasoning**:
The next step is to calculate sentiment scores using TextBlob and compare them with VADER scores and the original sentiment labels. This involves defining a function to get TextBlob sentiment, applying it to the cleaned tweets, labeling the sentiments, and then creating confusion matrices to compare the different sentiment classifications.



```python
from textblob import TextBlob

# Function to get TextBlob sentiment polarity
def get_textblob_sentiment(text):
    if isinstance(text, str):
        return TextBlob(text).sentiment.polarity
    else:
        return 0.0

# Apply TextBlob sentiment analysis to the cleaned tweets
df['TextBlob_Polarity'] = df['Cleaned_Tweet'].apply(get_textblob_sentiment)

# Create a new column for TextBlob sentiment labels
df['TextBlob_Sentiment_Label'] = df['TextBlob_Polarity'].apply(lambda score: 'Positive' if score >= 0.05 else ('Negative' if score <= -0.05 else 'Neutral'))

# Compare VADER and TextBlob sentiment labels
vader_textblob_comparison = pd.crosstab(df['Vader_Sentiment_Label'], df['TextBlob_Sentiment_Label'])
display("VADER vs TextBlob Sentiment Labels:")
display(vader_textblob_comparison)

# Compare original sentiment and VADER sentiment labels
original_vader_comparison = pd.crosstab(df['Sentiment'], df['Vader_Sentiment_Label'])
display("Original vs VADER Sentiment Labels:")
display(original_vader_comparison)

# Compare original sentiment and TextBlob sentiment labels
original_textblob_comparison = pd.crosstab(df['Sentiment'], df['TextBlob_Sentiment_Label'])
display("Original vs TextBlob Sentiment Labels:")
display(original_textblob_comparison)
```

## Filter by keyword

### Subtask:
Implement a function to filter the DataFrame based on a specific keyword in the 'Tweet' or 'Cleaned_Tweet' column. Display or visualize the sentiment of tweets containing the filtered keyword.


```python
def filter_tweets_by_keyword(df, keyword):
    """
    Filters the DataFrame to include tweets containing a specific keyword
    in either the 'Tweet' or 'Cleaned_Tweet' column (case-insensitive).

    Args:
        df: pandas DataFrame with 'Tweet' and 'Cleaned_Tweet' columns.
        keyword: The keyword to filter by.

    Returns:
        A filtered pandas DataFrame.
    """
    # Ensure keyword is a string
    keyword = str(keyword)

    # Filter rows where 'Tweet' or 'Cleaned_Tweet' contains the keyword (case-insensitive)
    filtered_df = df[
        df['Tweet'].str.contains(keyword, case=False, na=False) |
        df['Cleaned_Tweet'].str.contains(keyword, case=False, na=False)
    ].copy() # Use .copy() to avoid SettingWithCopyWarning

    return filtered_df

# Choose a keyword
keyword_to_filter = 'problem' # Example keyword

# Filter the DataFrame
filtered_tweets_df = filter_tweets_by_keyword(df, keyword_to_filter)

# Check if the filtered DataFrame is empty
if not filtered_tweets_df.empty:
    # Visualize the distribution of VADER sentiment labels for filtered tweets
    plt.figure(figsize=(8, 6))
    sns.countplot(data=filtered_tweets_df, x='Vader_Sentiment_Label', order=['Negative', 'Neutral', 'Positive'], palette='viridis', hue='Vader_Sentiment_Label', legend=False) # Added hue and legend=False
    plt.title(f'Distribution of VADER Sentiment for Tweets Containing "{keyword_to_filter}"', fontsize=14) # Enhanced title
    plt.xlabel('VADER Sentiment Label', fontsize=12) # Enhanced x-label
    plt.ylabel('Count', fontsize=12) # Enhanced y-label
    plt.show()
else:
    print(f"No tweets found containing the keyword: '{keyword_to_filter}'")
```
